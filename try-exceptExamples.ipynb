```python
#dividing numbers
num=24
try:
    num2=int(input("Enter your second num:"))
    num=num/num2
    print("success num=",num)
except:
    print("There was an error")
```

```python
#Except sepcific example
a = 15
try:
    b = int(input("Please enter a number to divide a"))
    a = a/b
    print("Success a=",a)
except ZeroDivisionError:
    print("The number you provided cant divide 15 because it is 0")
except ValueError:
    print("You did not provide a number")
except:
    print("Something went wrong")
        
```

```python
#Example of Except else and finally
x=10

try:
    y= int(input("Please enter a number to divide a"))
    x= x/y
except ZeroDivisionError:
    print("The number you provided cant divide 1 because it is 0")
except ValueError:
    print("You did not provide a number")
except:
    print("Something went wrong")
else:
    print("success x=",x)
finally:
    print("Completed")
```
