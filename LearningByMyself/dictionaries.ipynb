```python
#Dictionaries are a collection of key-value pairs. Each key is unique and maps to a value.
# Dictionaries are mutable, meaning you can change their content without changing their identity.
#Dictionaries are defined using curly braces {} and key-value pairs are separated by colons (:).
# Create the dictionary

Dict = {"key1": 1, "key2": "2", "key3": [3, 3, 3], "key4": (4, 4, 4), ('key5',): 5, (0, 1): 6}
print(Dict)
```

```python
# Access to the value by the key
Dict["key1"]
```

```python
 # Create a sample dictionary

sample_dict = {"Thriller": "1982", "Back in Black": "1980", \
                    "The Dark Side of the Moon": "1973", "The Bodyguard": "1992", \
                    "Bat Out of Hell": "1977", "Their Greatest Hits (1971-1975)": "1976", \
                    "Saturday Night Fever": "1977", "Rumours": "1977"}
print(sample_dict)
```

```python
#get the value of a specific key
print(sample_dict["Thriller"])
print(sample_dict["The Dark Side of the Moon"])
```

```python
# Get all the keys in dictionary

print(sample_dict.keys()) 
```

```python
#get all the values in dictinoary
print(sample_dict.values())
```

```python
# append a new key-value pair to the dictionary
sample_dict["Hotel California"] = "1976"
print(sample_dict)
```

```python
#delete a key-value pair from the dictionary
del(sample_dict["Back in Black"])
del(sample_dict["Rumours"])
print(sample_dict)
```

```python
# verify if the key is in dictinory or not
print("teen wolf" in sample_dict)
print("Thriller" in sample_dict)

```
