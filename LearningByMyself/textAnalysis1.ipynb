TEXT ANALYSIS-
Text analysis in python means examining text to find the useful information,patterns and meaning.

Using Python, we can find the most frequent words, count sentences, remove unnecessary words, and even estimate whether reviews are positive or negative.

PYTHON STRINGS FOR TEXT ANALYSIS
A string is a sequence of characters enclosed in quotes. Text analysis starts with strings because the data we want to analyze is text.

```python
text = "Python is easy and Python is powerful"
print(text)
```

1) Count Characters
Python's len() function counts the number of characters, including spaces.

```python
text="Python is high programming language"
print(len(text))
```

2) Convert text to lowercase and Uppercase
When analyzing text, Python, PYTHON, and python might need to be treated as the same word.

.lower() converts text to lowercase and .upper() into uppercase

```python
text1="PYTHon IS like ENglish"
print(text1.lower())
print(text1.upper())
```

3. Split text into words
.split() converts one string into a list of words.

```python
text2 = "Python is easy and Python is powerful"
words = text2.split()

print(words)
```

4. Count a particular word
Suppose we want to know how many times Python appears.

```python
text = "Python is easy and Python is powerful"

print(text.count("Python"))
```

```python
text = "Python is easy and PYTHON is powerful"
#But what if the text contains Python, PYTHON, and python then We can standardize it first.
text = text.lower()

print(text.count("python"))
```

5. Find the frequency of every word

```python
text = "python is easy and python is powerful "
words=text.split()

frequency = {}#set

for word in words:
    if word in frequency:
        frequency[word]+=1
    else:
        frequency [word]=1
print(frequency)
```
