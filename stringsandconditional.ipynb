```python
#STRING FUNCTIONS
#str.endwith("er.")---returns true if string ends with substr
str1="i'm studing python"
print(str1.endswith('hon'))
print(str1.endswith('ing'))

#str.capitalize()---capitalize 1st character
print(str1.capitalize())
#it is not making changes in original string, it's just printing with update

#str.replace(old,new)---replace all occurrences of old with new 
print(str1.replace("t","T"))
print(str1.replace("'m","am"))

#str.find(word)---returns 1st index of 1st occurrence
print("The word is at index:",str1.find("studing"))

#str.count("am")---#counts the occcurrence of substr in string
str2="hello i'm doremon from cartoon industry and i'm from japan"
print(str2.count("from"))
print(str2.count('o'))

#str.split()---splits the string into list of words
print(str2.split())

#str.upper()---converts all characters to uppercase
print(str2.upper())

#str.lower()---converts all characters to lowercase
print(str2.lower())

#str.strip()---removes all leading and trailing whitespaces
str3="   hello i'm doremon from cartoon industry and i'm from japan   "
print(str3.strip())
```

```python
#Grade of students based on marks
marks=int(input("Enter your marks:" ))
if(marks >= 90):
    grade='A'
elif(marks >= 80 and marks<90):
    grade='B'
elif(marks >= 70 and marks<80): 
    grade='C'
elif(marks >= 60 and marks<70): 
    grade='D'
else:
    grade='E'

print("Grade of the student:", grade)  
```
