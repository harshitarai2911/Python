EXCEPTIONS
Exceptions are alerts when something unexpected happens while running a program. It could be a mistake in the code or a situation that was not planned for. 

ERRORS VS EXCEPTIONS
 Errors are usually big problems that come from the computer or the system. They often make the program stop working completely.

 Exceptions are more like issues we can control. They happen because of something we did in our code and can usually be fixed, so the program keeps going.



Common Exceptions in Python
1) ZeroDivisionError
This error arises when an attempt is made to divide a number by zero. Division by zero is undefined in mathematics, causing an arithmetic error. 



```python
result=10/0
print(result)
# here ZeroDivisionError is found
```

2) ValueError
This error occurs when an inappropriate value is used within the code.

```python
num1=int("xyz")
print(num1)
#ValueError
```


3)FileNotFoundError
This exception is encountered when an attempt is made to access a file that does not exist.


```python
with open("file.txt", "r") as file:
        content = file.read()  
        
```

with -manages the file safely.

open() -opens the file.

"r" -specifies read mode for the file reading.

as file -stores the file object in a variable.

file.read()- reads the contents.

content- stores the resulting text.

IndexError
An IndexError occurs when an index is used to access an element in a list that is outside the valid index range.

```python
my_list = [1, 2, 3]
value = my_list[1] 
print(value)
missing = my_list[5]  # Raises IndexError
```

```python
dict={"fruit":"apple","vegetable":"potato"}
value=dict("mango")
```

AttributeError
An AttributeError occurs when an attribute or method is accessed on an object that doesn't possess that specific attribute or method. 



```python
text = "example"
length = len(text)  # No AttributeError, correct method usage
print(length)
missing = text.some_method()  
```

HANDLING EXCEPTIONS

Python has a handy tool called try and except that helps us manage exceptions.

```python
#using try-except
try:
    #attempting to divide 15 by 0
    result=15/0
except ZeroDivisionError:
        # Handling the ZeroDivisionError and printing an error message
    print("Error: Cannot divide by zero")
    # This line will be executed regardless of whether an exception occurred
print("outside of try and except block")
```
