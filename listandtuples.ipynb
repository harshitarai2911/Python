```python
#list is a built-in data structure in Python that is used to store multiple items in a single variable. 
# Lists are ordered, mutable (changeable), and allow duplicate elements. 
# They are defined by enclosing elements in square brackets [].
L = ["The Bodyguard", 7.0, 1992]
print(L)
print(type(L))

print(L[-1]) #accessing the last element of the list
print('the same element using negative and positive indexing:\n Postive:',L[1],
'\n Negative:' , L[-2]  )
print('the same element using negative and positive indexing:\n Postive:',L[2],
'\n Negative:' , L[-1]  )
```

```python
#LIST CONTENTS
#We can nest other lists, and we can also nest tuples and other data structures.
N=["The Bodyguard", 7.0, 1992, [1, 2], ("A", 1)]
print(N)
```

```python
#LIST OPERATIONS
L = ["The Bodyguard", 7.0, 1992]

#list slicing
print(L[0:2])

# Use extend to add elements to list
L.extend(['pop', 10])
print(L)

#Another similar method is append 
L.append("new")
print(L)
L.append(['a','b']) #nested list
print(L)

# Change the element based on the index
A = ["disco", 10, 1.2,"dance",34]
print('Before change:', A)
A[0] = 'hard rock'
print('After change:', A)

# Delete the element based on the index
del(A[0])
print(A)

# Split the string by comma
'A,B,C,D'.split(',')

```

```python
#COPY AND CLONE A LIST
# Copy (copy by reference) the list A
A = ["hard rock", 10, 1.2]
B = A
print('A:', A)
print('B:', B)

# Examine the copy by reference
print('B[0]:', B[0])
A[0] = "banana"
print('B[0]:', B[0])


# Clone (clone by value) the list A
B = A[:]
print('B:', B)
```

```python
#tuple is a built-in data structure in Python that is used to store multiple items in a single variable.
# Tuples are ordered, immutable (unchangeable), and allow duplicate elements.
# They are defined by enclosing elements in parentheses ().
tuple1 = ("disco",10,1.2 )
print(tuple1)
type(tuple1)
```

```python
#Indexing 
print(tuple1[0])
print(tuple1[1])
print(tuple1[2])

# Print the type of value on each index
print(type(tuple1[0]))
print(type(tuple1[1]))
print(type(tuple1[2]))
```

```python
tuple2 = ("radha", 25, 5.6,"richa", None)
print(tuple2)
print(tuple2[-1])
print(tuple2[-2])
print(tuple2[-3])
```

```python
#concatenation
tuple3 = tuple1 + tuple2
print(tuple3)
```

```python
#slicing
print(tuple2[0:3])
print(tuple3[1:4])

#length
len(tuple3)
print("length of tuple3 is:", len(tuple3))
```

```python
# sorting
marks = (87, 92 ,100, 94, 75, 91, 86)   
sortedmarks= sorted(marks) 
print("before sorting:", marks)
print("after sorting:", sortedmarks)
```

```python
#nested tuples
tuple4 = (1, 2, ("pop", 3, 4), (5, 6)) 
NestedT =(1, 2, ("pop", "rock") ,(3,4),("disco",(1,2)))
# Print element on each index

print("Element 0 of Tuple: ", NestedT[0])
print("Element 1 of Tuple: ", tuple4[1])
print("Element 2 of Tuple: ", tuple4[2])
print("Element 3 of Tuple: ", NestedT[3])
print("Element 4 of Tuple: ", NestedT[4])

# Print the second character of the second element in the first nested tuple
NestedT[2][1][1]
```
