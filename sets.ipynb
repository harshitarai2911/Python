```python
#Sets are unordered collection of unique elements. 
# They are mutable, meaning you can add or remove elements from a set after it has been created.
#  Sets are defined using curly braces {} or the set() constructor.

#creating a set
set1 = {1, 2, 3, 4, 5}
print(set1)  
set2 = {"apple", "banana", "cherry" , "apple"}  # duplicate "apple" will be ignored
print(set2)
```

```python
#converting a list to a set
list1= ['lion', 'tiger', 'bear', 'lion']
set1 = set(list1)
print(set1)

album_list = [ "Michael Jackson", "Thriller", 1982, "00:42:19", \
              "Pop, Rock, R&B", 46.0, 65, "30-Nov-82", None, 10.0]
album_set = set(album_list)             
print(album_set)

album_list= (["Barbie", "Bratz", "American Girl", "My Little Pony", "Barbie"])
album_set = set(album_list)
print(album_set)
```

```python
# Sample set

A = set(["Thriller", "Back in Black", "AC/DC"])
print(A)
```

```python
# Add element to set

A.add("Titanic")
print(A)
```

```python
# try to add a duplicate element
A.add("Titanic")  # This will not change the set since "Titanic" is already present
print(A)
```

```python
# Remove element from set
A.remove("Thriller")
print(A)
```

```python
 #Verify if the element is in the set

"AC/DC" in A
```

```python
# sets logic operations
album1 = set(["mai tera", "tum hi ho", "channa mereya"])
album2 = set(["dil jaaniye", "channa mereya", "dil diya gallan"])
print(album1, album2)

#print the union of two sets
print(album1 | album2) 
# we can also write print(album1.union(album2))

#print the intersection of two sets
print(album1 & album2)
#we can also write print(album1.intersection(album2))

#find the difference in album1 but not in album2
print(album1.difference(album2))
# we can also write print(album1 - album2)

#find the difference in album2 but not in album1
print(album2.difference(album1))   


```

```python
# check if subset (subsets are sets that contain only elements from another set)
print(album1.issubset(album2))
print(album2.issubset(album1))

# check if superset (supersets are sets that contain all elements of another set)
print(album1.issuperset(album2))
print(album2.issuperset(album1))
```
