# ARITHMETIC FUNCTIONS

```python
import numpy as np

# Creating an array
a = np.array([10, 20, 30, 40, 50])

# 1. Minimum
# Returns the smallest element
print("Minimum:", np.min(a))

# 2. Maximum
# Returns the largest element
print("Maximum:", np.max(a))

# 3. Mean
# Calculates the average of all elements
print("Mean:", np.mean(a))

# 4. Median
# Returns the middle value after sorting
print("Median:", np.median(a))

# 5. Sum
# Calculates the sum of all elements
print("Sum:", np.sum(a))

# 6. Variance
# Measures how far values are spread from the mean
print("Variance:", np.var(a))

# 7. Standard Deviation
# Measures the spread of data around the mean
print("Standard Deviation:", np.std(a))

# 8. All
# Returns True if all elements satisfy the condition
print("All:", np.all(a > 5))

# 9. Any
# Returns True if at least one element satisfies the condition
print("Any:", np.any(a > 40))
```

np.argmin() and np.argmax()

```python
import numpy as np
a = np.array([10, 5, 30, 2, 40])
# Returns the index of the minimum element
print("Minimum index:", np.argmin(a))
```

```python
import numpy as np
a = np.array([10, 5, 30, 2, 40])
# Returns the index of the maximum element
print("Maximum index:", np.argmax(a))
```

* In 2D for along column axis = 0 and for along row axis = 1

```python
import numpy as np
a = np.array([[10, 50, 30],[20, 5, 40]])

# Column-wise minimum index
print("Column min:", np.argmin(a, axis=0))

# Column-wise maximum index
print("Column max:", np.argmax(a, axis=0))

# Row-wise minimum index
print("Row min:", np.argmin(a, axis=1))

# Row-wise maximum index
print("Row max:", np.argmax(a, axis=1))
```

```python
# np.square(): square function multiplies each element by itself.
```

```python
import numpy as np
a = np.array([2, 3, 4, 5])
# Calculates square of each element
result = np.square(a)
print("Square:", result)
```

np.sin():sine function calculates the sine value of each element in an array.

NumPy's np.sin() takes angles in radians, not degrees. That's why we use np.deg2rad().

```python
import numpy as np
a = np.array([0, 30, 45, 90])
# Converting degrees into radians
radians = np.deg2rad(a)
# Calculates sine values
result = np.sin(radians)

print("Sine:", result)
```

np.cos(): The cosine function calculates the cosine value of each element in an array.

```python
import numpy as np
a = np.array([0, 30, 45, 60, 90])
# Converting degrees into radians
radians = np.deg2rad(a)

# Calculates cosine values
result = np.cos(radians)

print("Cosine:", result)
```

np.cumsum() calculates the cumulative sum of array elements.

It means each element is added to the sum of all previous elements.

```python
import numpy as np
a = np.array([10, 20, 30, 40, 50])
# Calculating cumulative sum
result = np.cumsum(a)
print("Cumulative Sum:", result)
```
