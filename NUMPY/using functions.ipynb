# CREATE NUMPY ARRAYS WITH RANDOM NUMBERS

1. rand()
- Generates a random number, typically a floating-point number between 0 and 1.
- In NumPy, the function is np.random.rand().

```python
import numpy as np
var = np.random.rand(4)
print(var)
```

```python
var1 = np.random.rand(2,5)
print(var1)
```

2. randn()
- Generates random numbers/values close to zero
- This may return positive as well negative no.
- The distribution has a mean of 0 and a standard deviation of 1.
- In NumPy, use np.random.randn().

```python
var2=np.random.randn(5)
print(var2)
```

3. randf()
- randf() is not a standard NumPy function or a built-in Python function.
- You may mean random() from Python's random module, which generates a floating-point number from 0 (inclusive) to 1 (exclusive).
interval[0.0,1.0]

```python

import random

print(random.random())
```

4. randint()
- Generates a random integer within a specified range.
- NumPy's np.random.randint(min,max,total_values) uses an exclusive upper limit.

```python
import numpy as np

print(np.random.randint(1, 10))
print(np.random.randint(10,50,5))
```
