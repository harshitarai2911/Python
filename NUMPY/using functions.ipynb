SPECIAL NUMPY ARRAY

ARRAY FILLED WITH 0'S

```python
import numpy as np
arr1=np.zeros(4)
print(arr1)
print()
arr2=np.zeros((3,4))
#3,4 representing rows and columns
print(arr2)
```

ARRAY FILLED WITH 1'S

```python
arr3=np.ones(5)
print(arr3)
```

CREATE AN EMPTY ARRAY

```python
x=np.empty(5)
print(x)
#same size ka jo bhi ishka previous array hai vo print kr dega
```

AN ARRAY WITH A RANGE OF ELEMENTS

```python
range_arr=np.arange(7)
print(range_arr)
```

ARRAY DIAGONAL ELEMMENT FILLED WITH 1'S

```python
arr_dia=np.eye(3)
print(arr_dia)
```

```python
arr_D=np.eye(3,5)
print(arr_D)
```
