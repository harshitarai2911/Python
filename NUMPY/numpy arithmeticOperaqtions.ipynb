# ARITHMETIC OPERATIONS

ADDITION

```python
import numpy as np
var=np.array([3,2,5,4,6])
varAdd=var+3
print(varAdd)
```

```python
var1=np.array([13,21,5,14,26])
var2=np.array([11,43,22,4,21])
varAdd=var1+var2
print(varAdd)
```

```python
# using function i.e np.add()
var1=np.array([13,21,5,14,26])
var2=np.array([11,43,22,4,21])
varAdd=np.add(var1,var2)
print(varAdd)
```

SUBTRACTION

```python
import numpy as np
var=np.array([3,2,5,4,6])
varSub=var-3
print(varSub)
```

```python
# using function i.e  np.subtract()
import numpy as np
a = np.array([10, 20, 30])
b = np.array([5, 10, 15])
result = np.subtract(a, b)
print(result)
```

MULTIPLICATION

```python
import numpy as np
var=np.array([3,2,5,4,6])
varMul=var*3
print(varMul)
```

```python
# using function i.e  np.multiply()
import numpy as np
a = np.array([2, 4, 6])
b = np.array([3, 5, 2])
result = np.multiply(a, b)
print(result)
```

DIVISION

```python
import numpy as np
var=np.array([3,7,12,24,61])
varDiv=var/3
print(varDiv)
```

```python
# using function i.e  np.divide()
import numpy as np
a = np.array([10, 20, 30])
b = np.array([2, 5, 10])
result = np.divide(a, b)
print(result)
```

MODULUS

```python
import numpy as np
var=np.array([3,2,5,4,6])
varMod=var%3
print(varMod)
```

```python
# using function i.e  np.mod()
import numpy as np
a = np.array([10, 20, 30, 40])
b = np.array([3, 6, 7, 9])
result = np.mod(a, b)
print(result)
```

RECIPROCAL

```python
import numpy as np
a = np.array([1, 2, 4, 5])
result = np.reciprocal(a)
print(result)

# We are getting zeros because the array contains integers, NumPy performs integer division.
```

```python
import numpy as np
b = np.array([1.0, 2.0, 4.0, 5.0])
result = np.reciprocal(b)
print(result)
```
