# Data type

```python
import numpy as np
var=np.array([3,6,1,9,7,2,0])
print("Data type:",var.dtype)
```

```python
var1=np.array([1.0,1.5,1.7,2.0])
print("Data type:",var1.dtype)
```

```python
var2=np.array(["A","G",1.7,2.0])
print("Data type:",var2.dtype)
#<U32 is a Unicode string data type, not a floating-point or integer data type.
```

```python
var3=np.array(["A","G","R","S"])
print("Data type:",var3.dtype)
#<U1 in NumPy means Unicode string with a maximum length of 1 character per element.
```

# DATATYPE CONVERSION

the list of NumPy data type shortcuts 

i-Integer
Integer numbers

u-Unsigned integer
Non-negative integers

f-Float
Decimal numbers

c-Complex
Complex numbers

b-Boolean
True or False

S-Bytes string
Byte characters

U-Unicode string
Text strings

O-Object
Python objects

M-Datetime
Date and time

m-Timedelta
Time duration

V-Void
Raw data / fixed-size memory

```python
x=np.array([1,2,0,9,8],dtype=np.int8)
print("Data type:",x.dtype)
print(x)
```

```python
x1=np.array([19,42,50,96,48],dtype="f")#f stands for float
print("Data type:",x1.dtype)
print(x)
```

Data conversion using a function

```python
var=np.array([3,6,1,9,7,2,0])
new=np.float32(var)#datatype conversion

#The the datatype in its previous form
new_one=np.int_(new)

print("Data type:",var.dtype)
print("Data type:",new.dtype)
print("Data type:",new_one.dtype)

print(var)
print(new)
print(new_one)
```

Direct conversion using astype()

```python
x3=np.array([19,42,50,96,48])
new_1=x3.astype(float)

print(x3)
print(new_1)
```
