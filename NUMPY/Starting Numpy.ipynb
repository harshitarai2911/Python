NumPy: A Python library for numerical computing.
Arrays: NumPy provides arrays to store and work with multiple values efficiently.

 Why use it? 
It makes mathematical operations on large collections of numbers easier and often faster than using ordinary Python lists.

Common uses: Mathematical calculations, scientific computing, data analysis, and machine learning.


```python
import numpy as np
x=np.array([3,7,4,5,9,0])
print(x)
print(type(x))
```

DIFFERENCE BETWEEN PYTHON LIST AND NUMPY ARRAY
Python List
- A list is a built-in Python data structure.
- It can store different data types, such as integers, strings, and floats.
- It is generally slower for large numerical calculations.
- It uses more memory for numerical data in many cases.
- Mathematical operations often require loops.
- Elements can be easily added or removed.
- Example: list1 = [1, 2, 3, 4]

NumPy Array
- A NumPy array is provided by the NumPy library.
- It usually stores elements of the same data type.
- It is generally faster for numerical calculations.
- It is often more memory-efficient for numerical data.
- It supports direct mathematical operations on all elements.
- Its size is typically fixed after creation, although new arrays can be created.
- Example: arr = np.array([1, 2, 3, 4])


%time (single line)
- It measures the execution time of one line of code.
- It starts witngle % symbol

%%time (Entire cell)
- It measures the execution time of the entire cell.
- It starts with two %% symbols.
- It must be written at the beginning of the cell.

```python
#python list
%time [j**4 for j in range(1,9)]
```

```python
#numpy array
import numpy as np
%time np.arange(1, 9)**4
```

```python
%%time
a = 10
b = 20
c = a + b
print(c)
```
