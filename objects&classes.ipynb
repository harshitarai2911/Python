OPPS IN PYTHON
OOPS (Object-Oriented Programming System) is a programming approach based on the concept of classes and objects. It helps organize code, improve reusability, and make programs easier to maintain.

```python
#Class-It is a blueprint for creating objects

#creating class
class Students:
    name="Ali"

#creating object(instant)
s1=Students()
print(s1)
print(s1.name)
s2=Students()
print(s2.name)


```

```python
class Cars:
    colour="black"
    brand="mahindra"
c1=Cars()
print(c1.colour)
print(c1.brand)
```

# CONSTRUCTOR
A constructor is a special method in a class that is automatically called when an object is created. It is used to initialize the object's data.

# In Python, the constructor is commonly written as __init__()
The constructor is called automatically when an object is created.

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

s1 = Student("Karan", 97)
print(s1.name)
print(s1.marks)

s2 = Student("Arjun", 88)
print(s2.name)
print(s2.marks)

```

# DEFAULT CONSTRUCTOR
A default constructor takes only self as an argument. It does not require any additional value from the user.

```python
class Student:
    def __init__(self):
        print("Student object created")

s1 = Student()
```

# PARAMETRIZED CONSTRUCTOR
A parameterized constructor accepts values as arguments when an object is created.

```python
class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("Rani")
print(s1.name)
s2 = Student("Raj")
print(s2.name)
```

# Default Parameterized Constructor
This constructor has parameters with default values. You can create an object with or without passing an argument.

```python
class Student:
    def __init__(self, name="Unknown"):
        self.name = name

s1 = Student()
s2 = Student("Preet")

print(s1.name)
print(s2.name)
```

```python
class Student:
    def __init__(self, name="Unknown"):
        self.name = name
        print("Student object created")

s1 = Student("shalu")
print(s1.name)

s2 = Student()
print(s2.name)
```

# Class & instance attributes
A class attribute is defined inside the class but outside the methods. It is shared by all instances of the class.

An instance attribute belongs to a specific object. It is usually defined inside the __init__() method using self.

```python

class Student:
    college = "ABC College"  # Class attribute

    def __init__(self, name, marks):
        self.name = name      # Instance attribute
        self.marks = marks    # Instance attribute

s1 = Student("Karan", 97)
s2 = Student("Arjun", 88)

print(s1.college)
print(s2.college)

print(s1.name, s1.marks)
print(s2.name, s2.marks)
```

# Method
A method is a function defined inside a class. It is used to perform a specific task related to that class or its objects.

```python
class Student:
    college_name="XYZ college"
    def __init__(self,name,rollno):
        self.name=name
        self.rollno=rollno
    #Method
    def Welcome(self):
        print("Welcome student",self.name)

    def attend(self):
        print("present:",self.rollno)

s1 = Student("Aru",7)
s2 = Student("Diya",14)
s1.Welcome()
s1.attend()
s2.Welcome()
s2.attend()
```
