OPEN FILE
f=open("file_name","mode")

```python
f=open("read&write.txt","r")
# r indicates to read mode
data=f.read()
#reading the entire data of file
print(data)
print(type(data))

f.close()
```

if we want to read the file and writing "r" , then by default txt file will open. 

```python
#reading particular set of characters
f=open("read&write.txt","r")
data=f.read(6)

print(data)
f.close()
```

```python
f=open("read&write.txt","r")
line1=f.readline()
print(line1)

line2=f.readline()
print(line2)

f.close()
```

An extra blank line appears after each printed line because readline() keeps the newline character (\n) and print() adds another newline.

r+ mode 
The r+ mode is used to read and write data in an existing file.

The file must already exist.

When writing, new data overwrites the existing characters from the current file pointer position.


```python
f = open("read&write.txt", "r+")
f.write("Hlo")
f.close()
```

w+ in file
It creates a file if it does not exist.

It truncates (erases) existing content when the file is opened.

```python
f = open("read&write.txt", "w+")

f.write("Hello")
print(f.read())
f.write("\n I'm learning python")
f.close()
```

a+ in file
Definition: The a+ mode is used to read and append data to a file. It creates a new file if it does not exist.

It creates a new file if the file does not exist.

It does not delete the existing content.

New data is always added at the end of the file.

The file pointer is initially positioned at the end for writing. To read from the beginning, use f.seek(0).



```python
f = open("read&write.txt", "a+")

f.write("\nWelcome to Python")

f.seek(0)
print(f.read())

f.close()
```

WITH SYNTAX
The with statement is used to open and handle a file automatically. It closes the file automatically after the block of code is executed, so we do not need to use f.close().



```python
with open("read&write.txt","r") as f:
    data = f.read()
    print(data)
```

Deleting a file
In Python, we use the os.remove() function to delete a file. First, we need to import the os module.

```python
import os

os.remove("read&write.txt")
```
