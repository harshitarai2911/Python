```python
#functions
#A function is a fundamental building block that encapsulates specific actions or compulations.
#Functions promote code modularity and reusability.
```

```python
#calculate total
def calculate_total(a, b):
    #def → tells Python that you are defining a function
    total = a+b
    return total
result = calculate_total(8,9)
print(result)
```

```python
# commonly used built in functions
# len()-Calculates the length of a sequence or collection
list=print(len([1,3,5,7,4,0]))

#len(): Calculates the length of a sequence or collection
total = print(sum([10, 20, 30, 40, 50]))

#max(): Returns the maximum value in an iterable
highest = print(max([5, 12, 8, 23, 16]))

#min(): Returns the minimum value in an iterable
lowest =print(min([5, 12, 8, 23, 16]))
```

```python
#Docstrings(Documentation Strings)
#Docstrings explain what a function does
#Placed inside triple quotes under the function definition
#It Helps other developers understand your function
def multiply(a, b):
    """
    This function multiplies two numbers.
    Input: a (number), b (number)
    Output: Product of a and b
    """
    print(a * b)
multiply(2,6)
```

```python
#SCOPES AND VARIABLES
#Global Scope: Variables defined outside functions; accessible everywhere
#Local Scope: Variables inside functions; only usable within that function
global_variable = "I'm global"
def example_function():
    local_variable = "I'm local"
    print(global_variable)  # Accessing global variable
    print(local_variable)   # Accessing local variable
example_function() # function call:As a result of this function call, it will print the values of the global and local variables within the function.
```

```python
#USING FUNCTIONS WITH LOOPS
def greet(name):
    return "Namashkara," + name
for _ in range(2):
    #In Python, _ is commonly used when you don't need the loop variable.
    # Python programmers conventionally use when the value is intentionally ignored.
    print (greet("Harshita"))
```

```python
#MODIFYING A LIST USING A FUNCTION
#Add an element i.e append()
def add_element(numbers):
    numbers.append(40)
    numbers.append(60)

my_list = [10, 20, 30]

add_element(my_list)

print(my_list)
```

```python
# to add more than one element at a time
#use extend()
def add_element(numbers):
    numbers.extend([30,65,43,78])
my_list = [10, 20, 30]
add_element(my_list)
print(my_list)
```

```python
# updating an element
def update_list(numbers):
    numbers[1] = 100

my_list = [10, 20, 30]
update_list(my_list)
print(my_list)
```

```python
#deleting an element
def delete_element(numbers):
    numbers.remove(20)

my_list = [10, 20, 30]

delete_element(my_list)

print(my_list)
```

```python
#Modifying a set using a function
def modify_set(numbers):
    numbers.update([40,50,60])
    numbers.remove(20)
my_set = {10, 20, 30}
modify_set(my_set)
print(my_set)

```
