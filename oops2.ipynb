Create student class that takes name & marks of 3 subjects as arguments in constructor.Then create a method to print the average

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def get_avg(self):
        sum=0
        for value in self.marks:
            sum+=value
        print("Hi",self.name,"your average marks is:",sum/3)

s1 = Student("Arun", [89, 94, 91])

print(s1.name)
print(s1.marks)
s1.get_avg()
#changing name
s1.name="Aadhya"
s1.get_avg()

```

STATIC METHODS
A static method is a method inside a class that does not use self or cls. It is used when a function is related to a class but does not need to access the object's or class's data.

We define a static method using the @staticmethod decorator.

```python
class Student:
    @staticmethod
    def hello():
        print("Hello, students!")

Student.hello()
```

```python
class Calculator:
    @staticmethod
    def add(a, b):
        print(a + b)

Calculator.add(10, 20)
```

```python
class Student:
    college_name = "ABC College"

    def __init__(self, name):
        self.name = name

    @staticmethod
    def is_valid_marks(marks):
        return 0 <= marks <= 100

s1 = Student("Rani")

print(s1.name)
print(Student.is_valid_marks(85))
print(Student.is_valid_marks(120))
```
